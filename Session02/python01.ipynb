```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True, dotenv_path="../.env.local")

my_api_key = os.getenv("OPENAI_API_KEY")

print(my_api_key[:5], my_api_key[-10:])
```

```python
client = OpenAI(api_key=my_api_key)

def call_llm(user_question):

    response = client.chat.completions.create(

        model="gpt-6-luna", #gpt-6-astra, gpt-6-sol 
        messages=[
            {"role": "system", "content": "You are an AI assistant. Give a thoughtful but concise response"},
            {"role": "user", "content": user_question}
        ]
    )

    print (response)
    print (f"Output: {response.choices[0].message.content}")
    return response.choices[0].message.content
```

```python
while True:
    
    user_question = input("Enter your question: ")
    if (user_question.lower() == "exit"):
        print("Exiting the chatbot.")
        break

    llm_response = call_llm(user_question)
    print("LLM response: " + llm_response)
```
